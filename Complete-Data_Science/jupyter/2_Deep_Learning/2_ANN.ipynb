```python
import numpy as np
import tensorflow as tf
```

```python
from tensorflow import keras
from tensorflow.keras.layers import Input,Dense
from tensorflow.keras.models import Model
```

```python
from sklearn.datasets import load_iris
from sklearn.compose import ColumnTransformer
```

set random seed for reprocubility

```python
np.random.seed(0)
tf.random.set_seed(0)
```

load iris datasets

```python
iris = load_iris()
```

Dividing the data in x and y

```python
X = iris.data
y = iris.target
```

process the data/standardize feature

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)   # X
```

one hot encode the label

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse=False)
```

```python
y_one_hot = encoder.fit_transform(y.reshape(-1,1))  # Y
```

split the data into train and test

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_scaled,y_one_hot,test_size=0.2,random_state=0)
```

defining the model using functional api

```python
inputs = Input(shape = (X_train.shape[1],))
```

```python
x = Dense(64,activation='relu')(inputs)
```

```python
x = Dense(32,activation='relu')(x)
```

```python
outputs = Dense(y_train.shape[1],activation='softmax')(x)
```

create the model

```python
model = Model(inputs = inputs,outputs = outputs)
```

Compile the model

```python
model.compile(optimizer='adam', loss='categorical_crossentropy',metrics=['accuracy'])
```

Train the model

```python
model.fit(X_train,y_train,epochs=30,batch_size=16,validation_data=(X_test,y_test))
```

Evaluate the model

```python
loss,accuracy = model.evaluate(X_test,y_test)
print(f"Test loss:{loss:.4f}")
print(f" Test accuracy: {accuracy:.4f}")
```
