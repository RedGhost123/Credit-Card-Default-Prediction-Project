```python
import keras
from keras.layers import Dense
from keras.layers import Input
from keras.layers import BatchNormalization
from keras.models import Sequential
```

```python
def build_model(hp):
 model = Sequential()
 model.add(Input(784,))
 model.add(Dense(units = hp.Int("units_1",min_value=200,max_value = 400,step=20),activztion=hp.Choice("activation_1",["relu","tanh"])))
 model.add(Dense(units = hp.Int("units_2",min_value=100,max_value = 200,step=20),activation=hp.Choice("activation_2",["relu","tanh"])))
 model.add(BatchNormalization())
 for layers in range(hp.Int("num_layers",2.4)):
    model.add(Dense(units=hp.Int("units_3",min_value = 50,max_value = 100,step_size =20),activation= hp.Choice("activation_3",["relu","tanh"])),kernal_regularizer=keras.regularizers.L2(l2=hp.Float("reg",min_value=0.001,max_value=0.1,sampling="log")))
 if hp.boolean("BatchNorm"):
    model.add(BatchNormalization())
 model.add(Dense(10,activation='softmax'))
 model.compile(loss="categorical_crossentropy",metrics = "Accurcy",optimizer=hp.Choice("optimizer",["adam","sgd"]))
 return model
```
