```python
class pwskills:
    def __init__(self,mentor):
        self.mentor = mentor
    def mentor_name(self):
        print(self.mentor)
```

```python
class data_science(pwskills):
    def __init__(self,mentor,mentor_email):
        self.mentor = mentor
        self.mentor_id = mentor_email
    def show_info(self):
        print(self.mentor,self.mentor_id)
```

```python
python_obj = data_science("vaibhav","vaibhav@gmail.com")
```

```python
python_obj.show_info()
```

```python
python_obj.mentor_name()
```

Super function

```python
class Pwskills:
    def __init__ (self,mentor):
        self.mentorname = mentor 
    def mentor_name(self):
        print(self.mentorname)

```

```python
class data_Science(Pwskills):
    def __init__(self,mentor,mentor_email):
        self.mentor = mentor
        super().__init__(mentor)
        self.mentor_mail_id = mentor_email
    def Showinfo(self):
        super().mentor_name()
        print(self.mentor,self.mentor_mail_id)
```

```python
python_obj2 = data_Science("aman","aman@gmail.com")
```

```python
python_obj2.Showinfo()
```
