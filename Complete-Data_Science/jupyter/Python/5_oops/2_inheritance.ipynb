Inheritance

```python
log.info("this is start of class a")
class a:
    try:
        def test(self):
           log.info("this is start of class a test1")
           print("this is class test1")
        def test2(self):
           log.info("this is start of test2")
           print("this is class test2")
    except Exception as e:
        print(e)
        log.info(e)
```

```python
log.info("this is class b")
class b(a):
    def test3(self):
        log.info("this is start of class b test3")
        print("this is class b test3")
    def test4(self):
        log.info("this is start of class b test4")
        print("this is class b test4")
```

```python
## Making Object of class
obj_a = a()
obj_b = b()
```

```python
log.info("calling class using obj_b")
log.info("calling test class")
obj_b.test()
obj_b.test2()
obj_b.test3()
obj_b.test4()
```

Multi-level inheritance

```python
class lecture:
    def topice(self):
        print("today class we are discussing about inheritance concept")
    def timing(self):
        print("timing for today class")
    def endtime(self):
        print("today i will try to end class early")
```

```python
class student(lecture):
    def stud_details(self):
        print("this class will give your student details")
```

```python
vaibhav = student()
```

```python
vaibhav.endtime()
vaibhav.stud_details()
vaibhav.timing()
vaibhav.topice()
```

```python
class teacher(student):
    def details(self):
        print("this method give you teacher details")
```

```python
final_object = teacher()
```

```python
final_object.stud_details()
```

```python
final_object.details()
```
