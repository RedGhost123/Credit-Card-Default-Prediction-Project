### New class (calculator)

```python
from functools import reduce


class calculator:
    
        def add(self,*args):
            return sum(args)
        def subtract(self,*args):
            return reduce(lambda a,b :a-b,args)
        def multiply(self,*args):
            return reduce(lambda a,b:a*b,args)
        def div(self,*args):
            return reduce(lambda a,b:a/b,args)
    
```

```python
cal = calculator()
```

```python
try:
    print(cal.subtract(6,7),cal.add(45,4),cal.multiply(9,9),cal.div(9,9))
except Exception as e:
    print(e)
```
