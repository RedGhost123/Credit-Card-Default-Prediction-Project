```python
import logging

# Configure logging
logging.basicConfig(
    filename="test_exception.txt",
    level=logging.INFO,
    format='%(asctime)s %(levelname)s - %(message)s'
)
```

```python
logging.info("this is start of my try block")
try:
    logging.info("asking user for input")
    age = int(input("enter your age"))
    logging.info("user has enter age = %d",age)
    if age < 0:
        raise ValueError("user has enter negative age")
        logging.info(ValueError)
    elif age < 18:
        raise Exception("user is underage")
        logging.info(Exception)
    else:
        print("valid user")
        logging.info("valid User")
except Exception as e:
    print(e)
    logging.info(e)
```
